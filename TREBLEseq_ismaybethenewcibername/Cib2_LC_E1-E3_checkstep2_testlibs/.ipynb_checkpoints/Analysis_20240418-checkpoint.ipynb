```python
E1v1 = pd.read_csv('RPTR/Staller_RPTR_E1_v1_MVS_0065_I1_CTATGGCGGA_CGGTAGAATT_S2.fastq.gz.assembled_output.csv')
E1v2 = pd.read_csv('RPTR/Staller_RPTR_E1_v2_MVS_0067_I1_GTCTTGTGCC_TCATGTAGGA_S4.fastq.gz.assembled_output.csv')
allstep2 = pd.concat([E1v1, E1v2])
allstep2_grouped = allstep2.groupby('BCs')['count'].sum()
allstep2_grouped
```

```python
test = 'AAAAAACCCAAACC'
allstep2[allstep2['BCs']==test]
```

```python
sns.histplot(E1v1, x='count', log_scale=True)
plt.title('V1 Primers: Reads histogram')
```

```python
sns.histplot(E1v2, x='count', log_scale=True)
plt.title('V2 Primers: Reads histogram')
```

```python
sns.histplot(allstep2_grouped, log_scale=True)
plt.title('Combined Reads histogram')
```
